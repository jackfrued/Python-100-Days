## 求解二次方程

```python
# 求解二次方程

## 初版回答：
## 这里用户要输入3个数字（），而且二次项系数不能为0。
try:
    del a,b
except NameError:
    print("变量已清空！")

s1 = input("请输入二次方程的参数a、b、c(以空格分开):")
## 还要确保用户输入的是数字，不对要重复输入
a,b,c = map(float,s1.split()) ## 不一定为int，可能为float，但是float包含int，所以用float最稳妥？
## 这里面的map是一个迭代器？
print(f"您输入的参数组成的二次方程为：{a}x^2 + {b}x + {c} = 0")

judge = b**2 - 4*a*c

if judge > 0:
    print("该方程有2个实根！")
elif judge == 0:
    print("该方程只有1个实根！")
else:
    print("该方程没有实根，但有两个负数根！")
```
