```python
# 求解二次方程

## 初版回答：
## 这里用户要输入3个数字（每个都要ValueError校验），而且二次项系数不能为0（业务校验）。
## 浮点数校验函数、
try:
    del a,b
except NameError:
    print("变量已清空！")

import cmath

def get_float_input(prompt:str) -> float:
    """
    获取用户输入的浮点数，并处理非法输入。
    :param prompt:用户输入提示信息
    :return：用户输入的浮点数
    """
    # 用户输错了要重新输入的
    # 这里面有两种错误需要校验，应该是逐一校验
    while True:
        s1 = input(prompt).strip()  ## 建议加个strip()先去一下空格
        try:
            return float(s1)
        except ValueError as e:
            print("请重新输入有效数字:",e)
            continue

def solve_quadratic(a:float,b:float,c:float) -> str:
    """
    求解二次方程的根
    """
    if a == 0:
        raise ValueError
    else:
        judge = b**2 - 4*a*c
        x1 = (-b - cmath.sqrt(judge)) / (2*a) ## 优先级有问题 / 和 * 是同一优先级
        x2 = (-b + cmath.sqrt(judge)) / (2*a)
        if judge > 0:
            return f"方程有两个‌不相等的实数根‌：{x1},{x2}"
        elif judge == 0:
            return f"方程有一个‌重根‌（两个相等的实数根）:,{x1}"
        else:
            return f"方程有两个‌共轭复数根‌‌：{x1},{x2}"

def main():
    while True:
        a = get_float_input("请输入二次项系数a:")
        if a == 0:
            a = get_float_input("请输入二次项系数a:") ## 这边如果第二次输入0的时候，没有第二次校验了，所以有问题
        else:
            break
    b = get_float_input("请输入一次项系数b:")
    c = get_float_input("请输入常数项系数c:")

    print(f"二次方程为{a}x^2 + {b}x + {c} = 0")
    print(f"二次方程的解为：\n",solve_quadratic(a,b,c))
    
main()
```

```python
# 修正版本
import cmath
import math

def get_float_input(prompt: str) -> float:
    while True:
        s = input(prompt).strip()
        try:
            x = float(s) # float()可以解析更多数值字符串,3、3.14、1e-3
        except ValueError as e:
            print("请重新输入有效数字：", e)
            continue
        if not math.isfinite(x): # float()还是会接受"nan", "inf", "-inf" 也能转成功（结果是非有限数）
            print("不接受 inf / -inf / nan，请重新输入。")
            continue
        return x

def solve_quadratic(a: float, b: float, c: float) -> str:
    """求解二次方程的根（允许复根）"""
    if math.isclose(a,0):
        raise ValueError("a 不能为 0（否则不是二次方程）") # 扩充一下错误信息

    judge = b**2 - 4*a*c
    sqrt_d = cmath.sqrt(judge) # 多个变量，减少cmath.sqrt()重算一次

    x1 = (-b - sqrt_d) / (2 * a)
    x2 = (-b + sqrt_d) / (2 * a)

    eps = 1e-12  # 容差，是一个极小的正数
    if judge > eps:  # 大于容差，就可以默认为大于0了
        return f"方程有两个不相等的实数根：{x1}, {x2}"
    elif abs(judge) <= eps: # 绝对值小于容差，就可以默认为等于0
        return f"方程有一个重根（两个相等的实数根）：{x1}"
    else:
        return f"方程有两个共轭复数根：{x1}, {x2}"

def main():
    while True: # while循环要搞清楚
        a = get_float_input("请输入二次项系数 a: ")
        if a != 0:
            break
        print("a 不能为 0，请重新输入。")

    b = get_float_input("请输入一次项系数 b: ")
    c = get_float_input("请输入常数项 c: ")

    print(f"二次方程为：{a}x^2 + {b}x + {c} = 0")
    print("二次方程的解为：")
    print(solve_quadratic(a, b, c))

main()
```

## 计算三角形的面积

```python
# 通过用户输入三角形三边长度，并计算三角形的面积：
# 先判断是否为三角形：两边之和大于第三遍
# 再求三角形面积：海伦公式求解

## 初始版本

import math

while True:
    s = input("请输入三角形的三边长（用,分开,依次为a,b,c）：")
    try:
        a,b,c = map(float,s.strip().split(","))
        # 值合法性校验
        if not all(map(math.isfinite,(a,b,c))):
            raise ValueError("请不要输入nan -fin fin。")
        # 业务规则校验
        if ((a+b>c) and (a+c>b) and (b+c>a)):
            print("可以构成三角形！")
        else:
            print("无法构成三角形，请重新输入！")
            raise ValueError("任意两边之和没有大于第三边。")
        break
    except ValueError:
        print("请重新输入正确的格式")

p = (a+b+c) / 2
area_of_triangle = (p*(p-a)*(p-b)*(p-c))**0.5

print(f"输入三角形的各边长为{a},{b},{c},三角形面积为{area_of_triangle:0.2f}!")

```

## 判断字符串是否为数字

```python
# 自定义一个是否为数字的函数

def is_number(s):
    try:
        float(s)
        return True
    except ValueError:
        pass
    
    try:
        import unicodedata
        unicodedata.numeric(s)
        return True
    except (ValueError,TypeError):
        pass

    return False

print(is_number("1"))
print(is_number("3.5    "))
print(is_number("六"))
print(is_number("九   "))

```

## 约瑟夫环

```python
## 约瑟夫生死者小游戏
# 30 个人在一条船上，超载，需要 15 人下船。
# 于是人们排成一队，排队的位置即为他们的编号。
# 报数，从 1 开始，数到 9 的人下船。
# 如此循环，直到船上仅剩 15 人为止，问都有哪些编号的人下船了呢？

number_of_boat = [x for x in range(1,31)]

out_of_boat = []

idx = 0

while len(out_of_boat)  <= 15:
    idx = (idx + 9 -1) % len(number_of_boat)
    # 每次根据船上人数重算索引
    # 要按照相对顺序去删除
    drop = number_of_boat.pop(idx)
    print(f"{drop}号下船了！")
    out_of_boat.append(drop)

print(out_of_boat)

```

```python
## 约瑟夫生死者小游戏
## 用collection.deque旋转
from collections import deque

people = deque(range(1,31))
k = 9

out = []

while len(out) < 15:
    people.rotate(-(k-1)) # 把第9个人换到最前面，然后用popleft删除
    out.append(people.popleft())

print(out)
```

```python
## 五人分鱼
# A、B、C、D、E 五人在某天夜里合伙去捕鱼，到第二天凌晨时都疲惫不堪，于是各自找地方睡觉。
# 日上三杆，A 第一个醒来，他将鱼分为五份，把多余的一条鱼扔掉，拿走自己的一份。
# B 第二个醒来，也将鱼分为五份，把多余的一条鱼扔掉拿走自己的一份。 。
# C、D、E依次醒来，也按同样的方法拿鱼。
# 问他们至少捕了多少条鱼?

def split_fish(after_num : int):
    if after_num % 4 == 0 :
        # 想判断before_num是不是整数，可以向前判断after_num是不是整数
        return (after_num / 4 * 5 +1)
    else:
        return None

num = 0 # 最后一个人E分到的鱼的数量，循环内依次加1

while True:
    num += 1 # 依次校验1,2,3,4，……行不行
    x = num # 每次校验的数据，如果都通过校验，x的值就是总鱼数
    ok = True # 先定义，防止内层定义导致首次到外层时报错变量未定义
    for i in range(1,6): # 校验就是累计算5次，都是整数
        if split_fish(x) is not None: # 鱼儿是整数，不是整数就继续下一次校验
            x = split_fish(x)
        else:
            ok = False # 
            break
    if ok: # 这个标志变量，可以协助跳出外层循环
       break 
    
print(f"已找到x,x的值为{x}!")
```
