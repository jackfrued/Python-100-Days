## 求解二次方程

```python
# 求解二次方程

## 初版回答：
## 这里用户要输入3个数字（每个都要ValueError校验），而且二次项系数不能为0（业务校验）。
## 浮点数校验函数、
try:
    del a,b
except NameError:
    print("变量已清空！")

import cmath

def get_float_input(prompt:str) -> float:
    """
    获取用户输入的浮点数，并处理非法输入。
    :param prompt:用户输入提示信息
    :return：用户输入的浮点数
    """
    # 用户输错了要重新输入的
    # 这里面有两种错误需要校验，应该是逐一校验
    while True:
        s1 = input(prompt).strip()  ## 建议加个strip()先去一下空格
        try:
            return float(s1)
        except ValueError as e:
            print("请重新输入有效数字:",e)
            continue

def solve_quadratic(a:float,b:float,c:float) -> str:
    """
    求解二次方程的根
    """
    if a == 0:
        raise ValueError
    else:
        judge = b**2 - 4*a*c
        x1 = (-b - cmath.sqrt(judge)) / (2*a) ## 优先级有问题 / 和 * 是同一优先级
        x2 = (-b + cmath.sqrt(judge)) / (2*a)
        if judge > 0:
            return f"方程有两个‌不相等的实数根‌：{x1},{x2}"
        elif judge == 0:
            return f"方程有一个‌重根‌（两个相等的实数根）:,{x1}"
        else:
            return f"方程有两个‌共轭复数根‌‌：{x1},{x2}"

def main():
    while True:
        a = get_float_input("请输入二次项系数a:")
        if a == 0:
            a = get_float_input("请输入二次项系数a:") ## 这边如果第二次输入0的时候，没有第二次校验了，所以有问题
        else:
            break
    b = get_float_input("请输入一次项系数b:")
    c = get_float_input("请输入常数项系数c:")

    print(f"二次方程为{a}x^2 + {b}x + {c} = 0")
    print(f"二次方程的解为：\n",solve_quadratic(a,b,c))
    
main()
```

```python
# 修正版本
import cmath
import math

def get_float_input(prompt: str) -> float:
    while True:
        s = input(prompt).strip()
        try:
            x = float(s) # float()可以解析更多数值字符串,3、3.14、1e-3
        except ValueError as e:
            print("请重新输入有效数字：", e)
            continue
        if not math.isfinite(x): # float()还是会接受"nan", "inf", "-inf" 也能转成功（结果是非有限数）
            print("不接受 inf / -inf / nan，请重新输入。")
            continue
        return x

def solve_quadratic(a: float, b: float, c: float) -> str:
    """求解二次方程的根（允许复根）"""
    if math.isclose(a,0):
        raise ValueError("a 不能为 0（否则不是二次方程）") # 扩充一下错误信息

    judge = b**2 - 4*a*c
    sqrt_d = cmath.sqrt(judge) # 多个变量，减少cmath.sqrt()重算一次

    x1 = (-b - sqrt_d) / (2 * a)
    x2 = (-b + sqrt_d) / (2 * a)

    eps = 1e-12  # 容差，是一个极小的正数
    if judge > eps:  # 大于容差，就可以默认为大于0了
        return f"方程有两个不相等的实数根：{x1}, {x2}"
    elif abs(judge) <= eps: # 绝对值小于容差，就可以默认为等于0
        return f"方程有一个重根（两个相等的实数根）：{x1}"
    else:
        return f"方程有两个共轭复数根：{x1}, {x2}"

def main():
    while True: # while循环要搞清楚
        a = get_float_input("请输入二次项系数 a: ")
        if a != 0:
            break
        print("a 不能为 0，请重新输入。")

    b = get_float_input("请输入一次项系数 b: ")
    c = get_float_input("请输入常数项 c: ")

    print(f"二次方程为：{a}x^2 + {b}x + {c} = 0")
    print("二次方程的解为：")
    print(solve_quadratic(a, b, c))

main()
```

## 计算三角形的面积

```python
# 通过用户输入三角形三边长度，并计算三角形的面积：
# 先判断是否为三角形：两边之和大于第三遍
# 再求三角形面积：海伦公式求解

## 初始版本

import math

while True:
    s = input("请输入三角形的三边长（用,分开,依次为a,b,c）：")
    try:
        a,b,c = map(float,s.strip().split(","))
        # 值合法性校验
        if not all(map(math.isfinite,(a,b,c))):
            raise ValueError("请不要输入nan -fin fin。")
        # 业务规则校验
        if ((a+b>c) and (a+c>b) and (b+c>a)):
            print("可以构成三角形！")
        else:
            print("无法构成三角形，请重新输入！")
            raise ValueError("任意两边之和没有大于第三边。")
        break
    except ValueError:
        print("请重新输入正确的格式")

p = (a+b+c) / 2
area_of_triangle = (p*(p-a)*(p-b)*(p-c))**0.5

print(f"输入三角形的各边长为{a},{b},{c},三角形面积为{area_of_triangle:0.2f}!")

```

## 判断字符串是否为数字

```python
# 自定义一个是否为数字的函数

def is_number(s):
    try:
        float(s)
        return True
    except ValueError:
        pass
    
    try:
        import unicodedata
        unicodedata.numeric(s)
        return True
    except (ValueError,TypeError):
        pass

    return False

print(is_number("1"))
print(is_number("3.5    "))
print(is_number("六"))
print(is_number("九   "))

```

## 判断闰年

```python
# 闰年
# 闰年规则
# ‌能被4整除但不能被100整除‌（比如2024年）。
# ‌能被400整除‌（比如2000年）。
# 如果既不能被4整除，或者能被100整除但不能被400整除（比如1900年），那就是平年。

year = int(input("输入一个年份："))

if (year % 4 == 0 and year % 100 != 0) or (year % 400 == 0):
    print(f"{year} 是闰年")
else:
    print(f"{year}不是闰年")
```

## 判断质数

```python
# 质数：大于1，因数只有1和自身
## 初始版本
import math

n = int(input("输入一个正整数："))

def is_prime(n:int) -> bool:
    for i in range(2,int(math.sqrt(n))+1):
        if n == 2 or n == 3:
            return True
        elif n % i == 0:
            return False
    return True
print(f"{n}是质数吗？：", is_prime(n))

```

```python
# 修正版本
import math

n = int(input("输入一个正整数："))

def is_prime(n: int) -> bool:
    if n <= 1:
        return False
    if n <= 3:  # 2 和 3
        return True
    if n % 2 == 0: # 偶数肯定不行
        return False

    limit = math.isqrt(n)
    for i in range(3, limit + 1, 2):
        if n % i == 0:
            return False
    return True

print(f"{n}是质数吗？：", is_prime(n))
```

## 斐波那契数列

```python
# 斐波那契数列
# 0,1,1,2,3,5  后一项等于前两项的和

# 递归实现，占用大量资源
def fibo(n):
    if n == 0:
        return 0
    if n == 1:
        return 1
    if n > 1:
        return fibo(n-1) + fibo(n-2)

for i in range(0,100):
    print(fibo(i), end = " ") 



```

```python
# 循环实现
def fibo2(n):
    if n == 0 or n == 1 :
        s = n
    if n > 1:
        count = 1
        s2 = 0 # 往前第二个，初始化值不要放到循环体里面
        s1 = 1 # 往前第一个
        while count < n:
            s = s1 + s2
            s1,s2 = s,s1
            count += 1
    return s

for i in range(10):
    print(fibo2(i))
```

## 阿姆斯特朗数

```python
## 如果一个 n 位正整数等于其各位数字的 n 次方之和，则称该数为阿姆斯特朗数。 例如 1^3 + 5^3 + 3^3 = 153。

## 输出 1000 以内的阿姆斯特朗数： 1, 2, 3, 4, 5, 6, 7, 8, 9, 153, 370, 371, 407。

def amo(n:int) -> list:
    """
    
    """
    result = []
    for i in range(1,n+1):
        n = len(str(i))

        sum = 0
        temp = i
        while temp > 0:
          digit = temp % 10
          sum += digit ** n
          temp //= 10
        
        if i == sum:
            result.append(i)
    
    return result

            
print(amo(1000))
```
